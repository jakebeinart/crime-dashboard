```python
import pandas as pd
import folium
from folium.plugins import HeatMap

crime_data = pd.read_csv('../data/raw_2021-2024.csv')

crime_data = crime_data.dropna(subset=['Latitude', 'Longitude'])
```

```python

heat_data = [[row['Latitude'], row['Longitude']] for index, row in crime_data.iterrows()]
```

```python
m = folium.Map(location=[38.6270, -90.1994], zoom_start=12)
```

```python
HeatMap(heat_data).add_to(m)

m.save('crime_heatmap.html')
```

```python
m = folium.Map(location=[38.6270, -90.1994], zoom_start=12)

# filter by felony
f_crime_data = crime_data[crime_data['FelMisdCit'] == 'F']

f_heat_data = [[row['Latitude'], row['Longitude']] for index, row in f_crime_data.iterrows()]

HeatMap(f_heat_data).add_to(m)

m.save('f_crime_heatmap.html')
```

```python
m = folium.Map(location=[38.6270, -90.1994], zoom_start=12)

# filter by firearm
gun_crime_data = crime_data[crime_data['FirearmUsed'] == 'Yes']

gun_heat_data = [[row['Latitude'], row['Longitude']] for index, row in gun_crime_data.iterrows()]

HeatMap(gun_heat_data).add_to(m)

m.save('gun_crime_heatmap.html')
```
