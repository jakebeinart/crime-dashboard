```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# CONCATENATE NEW DATA

crime_data = pd.read_csv('../data/raw_2021-2024.csv')
crime_data_jan = pd.read_csv('../data/January2025.csv')

raw_total_data = pd.concat([crime_data, crime_data_jan])

# raw_total_data['IncidentNum'] = raw_total_data['IncidentNum'].str.strip()



raw_total_data.to_csv('raw_2021-jan2025.csv', index=False)

```

```python
# REMOVE UPDATES

raw_total_data["LastSuppDate"] = pd.to_datetime(raw_total_data["LastSuppDate"], format="mixed", dayfirst=False)

# Sort the DataFrame to ensure the latest records appear last
raw_total_data_sorted = raw_total_data.sort_values(by=["IncidentNum", "Offense", "LastSuppDate"], ascending=[True, True, False])

# Drop duplicates, keeping only the latest entry per IncidentNum & Offense
raw_total_data_sorted_cleaned = raw_total_data_sorted.drop_duplicates(subset=["IncidentNum", "Offense"], keep="first")

# Save the cleaned data
raw_total_data_sorted_cleaned.to_csv("cleaned_nibrs_data.csv", index=False)

print("Outdated records removed. Cleaned data saved.")
```

```python
# Group by IncidentDate, count "MURDER 1ST DEGREE" offenses, and sort in descending order
murder_counts = (
    raw_total_data_sorted_cleaned[raw_total_data_sorted_cleaned["Offense"] == "MURDER 1ST DEGREE"]
    .groupby("IncidentDate")
    .size()
    .reset_index(name="Murder_Count")
    .sort_values(by="Murder_Count", ascending=False)
)

# Display the result
print(murder_counts)
```
